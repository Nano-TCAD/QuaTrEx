```python
import numpy as np
num_kpoints = np.array([1, 3, 1])
nkpts = np.prod(num_kpoints)
ind_mat = np.roll(np.arange(nkpts, dtype=np.int32), -(nkpts//2)).reshape(num_kpoints)

ki = 1
kip = 0

mip = np.array(np.where(ind_mat == kip))
mi = np.array(np.where(ind_mat == ki))
md = tuple(mi-mip)
mmd = tuple(mip-mi)
print(int(ind_mat[md]))
# minus ekd
print(ind_mat[mmd])
```

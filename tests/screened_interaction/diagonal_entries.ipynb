```python
import numpy as np

PATH = "/usr/scratch/bucaramanga/awinka/quatrex_results/testing/diag_screened_interaction/"
num_orbs = 792
num_energy = 512
wg_diag = np.zeros((num_orbs, num_energy), dtype=np.complex128)
wl_diag = np.zeros((num_orbs, num_energy), dtype=np.complex128)
wr_diag = np.zeros((num_orbs, num_energy), dtype=np.complex128)
for i in range(num_orbs):
    diag_el = "_gw2s_"+str(i)+"x"+str(i)+"_iter0.npy"
    wg_diag[i] = np.load(PATH+"wg"+diag_el)
    wl_diag[i] = np.load(PATH+"wl"+diag_el)
    wr_diag[i] = np.load(PATH+"wr"+diag_el)
```

```python
print(np.imag(wg_diag-wl_diag).max())
print(np.imag(wr_diag).max())
print(np.imag(wg_diag-wl_diag).min())
print(np.imag(wr_diag).min())
```

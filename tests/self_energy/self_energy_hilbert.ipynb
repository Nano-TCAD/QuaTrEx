```python
import numpy as np

scratch_path = '/usr/scratch/bucaramanga/awinka/quatrex_results/self_energy_data/'

sg = np.load(scratch_path + 'sg_h2g_iter1_e-10.0_kp0.npy')
sl = np.load(scratch_path + 'sl_h2g_iter1_e-10.0_kp0.npy')
sr = np.load(scratch_path + 'sr_h2g_iter1_e-10.0_kp0.npy')
```

```python
spec1 = sg-sl
spec2 = sr-sr.T.conj()
```

```python
print(spec1[0:3, 0:3]/2)
print(spec2[0:3, 0:3])
np.allclose(np.imag(spec1), np.imag(spec2)*2, rtol=1e-2)
```

```python
import numpy as np

# Convolution testing
nzeros1 = 8
nzeros2 = 0
x1 = np.concatenate((np.zeros(nzeros1), np.array([1, 2, 3, 4, 5, 6, 7 , 8]), np.zeros(nzeros2)))
x2 = x1[::-1]  #np.array([1, 2, 3, 4, 5])
np.convolve(x1, x2)
```

```python
x1_t = np.fft.fft(x1, n=2*len(x1)-1)
x2_t = np.fft.fft(x2, n=2*len(x2)-1)
np.real(np.round(np.fft.ifft(x1_t*x2_t)))
```
